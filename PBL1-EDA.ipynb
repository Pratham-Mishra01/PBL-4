```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file_path = "Stellar Classification Dataset 1.csv"  # Ensure the file is in the same directory or provide full path
df = pd.read_csv(file_path)

# Drop the first row if it contains NaN values (misalignment issue)
df_cleaned = df.dropna().reset_index(drop=True)

# Select only numerical columns
numerical_cols = df_cleaned.select_dtypes(include=['float64']).columns

# Compute correlation matrix
correlation_matrix = df_cleaned[numerical_cols].corr()

# Plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Heatmap of Correlations Between Numerical Features")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
file_path = r"C:\Users\prath\OneDrive\Desktop\Pandas\Stellar Classification Dataset.csv"
df = pd.read_csv(file_path)

# Drop the first row if it contains NaN values (misalignment issue)
df_cleaned = df.dropna().reset_index(drop=True)

# Select only numerical columns
numerical_cols = df_cleaned.select_dtypes(include=['float64']).columns

# Compute correlation matrix
correlation_matrix = df_cleaned[numerical_cols].corr()

# Plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Heatmap of Correlations Between Numerical Features")
plt.show()
```

```python
df_cleaned.head(100)
```

```python
df_cleaned["type"] = df_cleaned["type"].astype(str)

# Now apply string operations safely
df_cleaned["type"] = df_cleaned["type"].str.strip().str.lower()

# Define mapping
galaxy_mapping = {
    "spiral galaxy": 0,
    "barred spiral galaxy": 1,
    "elliptical galaxy": 4,
    "lenticular galaxy": 2,
    "dwarf galaxy": 3
}

# Apply mapping
df_cleaned["type"] = df_cleaned["type"].map(galaxy_mapping)

# Verify if mapping worked
print(df_cleaned["type"].value_counts())
```

```python
df_cleaned.head(100)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
min_max_scaler = MinMaxScaler()
df_cleaned[['u', 'g', 'r', 'i', 'z']] = min_max_scaler.fit_transform(df_cleaned[['u', 'g', 'r', 'i', 'z']])
```

```python
z_score_scaler = StandardScaler()
df_cleaned['redshift'] = z_score_scaler.fit_transform(df_cleaned[['redshift']])
```
